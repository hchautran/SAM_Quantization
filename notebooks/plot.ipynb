```python
import matplotlib.pyplot as plt
import numpy as np

def plot_entropy_by_block(entropy_data, figsize=(15, 10)):
    """
    Plot entropy mean and variance for each attention block.
    
    Args:
        entropy_data: Dictionary with block names as keys and list of tuples as values
                     Each tuple contains (head_index, mean, variance)
        figsize: Figure size tuple (width, height)
    """
    # Calculate number of subplots needed
    n_blocks = len(entropy_data)
    n_cols = 4  # 4 plots per row
    n_rows = (n_blocks + n_cols - 1) // n_cols
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize)
    axes = axes.flatten() if n_blocks > 1 else [axes]
    
    for idx, (block_name, data) in enumerate(entropy_data.items()):
        # Extract data from tuples
        head_indices = [item[0] for item in data]
        means = [item[1] for item in data]
        variances = [item[2] for item in data]
        
        # Calculate standard deviation from variance for error bars
        std_devs = [np.sqrt(var) for var in variances]
        
        # Create the plot
        ax = axes[idx]
        bars = ax.bar(head_indices, means, yerr=std_devs, capsize=5, 
                     alpha=0.7, color='skyblue', edgecolor='navy', linewidth=0.5)
        
        # Customize the plot
        ax.set_xlabel('Head Index')
        ax.set_ylabel('Entropy Mean')
        ax.set_title(f'{block_name.replace("image_encoder.blocks.", "Block ").replace(".attn", "")}')
        ax.grid(True, alpha=0.3)
        
        # Set x-axis ticks to show all head indices
        ax.set_xticks(head_indices)
        
        # Add value labels on top of bars (optional)
        for bar, mean in zip(bars, means):
            height = bar.get_height()
            ax.text(bar.get_x() + bar.get_width()/2., height + max(std_devs)*0.01,
                   f'{mean:.2f}', ha='center', va='bottom', fontsize=8, rotation=45)
    
    # Hide empty subplots
    for idx in range(n_blocks, len(axes)):
        axes[idx].set_visible(False)
    
    plt.tight_layout()
    plt.show()

# Usage
plot_entropy_by_block(entropy_mean_variance)
```

```python
import torch

# Parameters
B = 2  # batch size
H, W = 1, 1  # height, width (so H*W = 1)
num_heads = 3
C = 1  # channels per head

# Create example tensor with clear values to track
# Shape: (B, H*W, 3, num_heads, C)
qkv = torch.arange(B * 1 * 3 * num_heads * C).reshape(B, 1, 3, num_heads, C).float()

print("Original qkv shape:", qkv.shape)
print("Original qkv:")
print(qkv)
print()

# Step 1: permute(2, 0, 3, 1, 4) - moves dimensions around
# From: (B, H*W, 3, num_heads, C) = (0, 1, 2, 3, 4)
# To:   (3, B, num_heads, H*W, C) = (2, 0, 3, 1, 4)
qkv_permuted = qkv.permute(2, 0, 3, 1, 4)
print("After permute - shape:", qkv_permuted.shape)
print("After permute:")
print(qkv_permuted)
print()

# Step 2: reshape to (3, B*num_heads, H*W, C)
qkv_reshaped = qkv_permuted.reshape(3, B * num_heads, 1, C)
print("After reshape - shape:", qkv_reshaped.shape)
print("After reshape:")
print(qkv_reshaped)
print()

# Let's see how B*num_heads is organized
print("Focus on how B*num_heads dimension is organized:")
print("B=0, heads 0,1,2 then B=1, heads 0,1,2")
for i in range(B * num_heads):
    batch_idx = i // num_heads
    head_idx = i % num_heads
    print(f"Position {i}: B{batch_idx}, head{head_idx} = {qkv_reshaped[0, i, 0, 0].item()}")
```
